## 6. Benchmark Evaluation on Held-Out Test Set
Computes official SacreBLEU (1..4), chrF++, and Word Error Rate (WER).


```python
import os, sys, json, numpy as np
for p in ['.', '/content']:
    if p not in sys.path:
        sys.path.insert(0, p)

from ml.inference.predictor import ISLPredictor
from ml.training.metrics import compute_bleu, compute_chrf, compute_wer

predictor = ISLPredictor(checkpoint_path='ml/checkpoints/isl_transformer_best.pt')

with open('data/manifests/test_manifest.json', 'r', encoding='utf-8') as f:
    test_samples = json.load(f)

hypotheses = []
references = []

for sample in test_samples:
    if sample.get('pose_path') and os.path.exists(sample['pose_path']):
        feat = np.load(sample['pose_path'])
        res = predictor.predict_pose_sequence(feat)
        hypotheses.append(res['english_text'])
    else:
        hypotheses.append('')
    references.append(sample['english_text'])

bleu = compute_bleu(hypotheses, references)
chrf = compute_chrf(hypotheses, references)
wer = compute_wer(hypotheses, references)

print('='*50)
print('🎯 TEST BENCHMARK RESULTS (HELD-OUT UNSEEN SIGNERS)')
print('='*50)
for k, v in bleu.items():
    print(f'{k.upper()}: {v}')
print(f'chrF++: {chrf}')
print(f'WER:    {wer}%')
print('='*50)
```

## 7. Interactive Translation & Audio Playback Demo
Generates English and native Telugu script (`తెలుగు`) with playable neural speech synthesis.


```python
from backend.app.translation.translator import EnglishToTeluguTranslator
from backend.app.tts.engine import TTSEngine
from IPython.display import Audio, display

translator = EnglishToTeluguTranslator()
tts = TTSEngine()

sample_english = 'Where is the hospital?'
telugu_output = translator.translate(sample_english)

print('📖 English Prediction:', sample_english)
print('📖 Telugu Translation (తెలుగు):', telugu_output.telugu_text)

# Synthesize speech directly to MP3 files for Colab audio player
from gtts import gTTS
tts_en = gTTS(text=sample_english, lang='en')
tts_en.save('speech_en.mp3')
print('\n🔊 English Speech:')
display(Audio('speech_en.mp3'))

tts_te = gTTS(text=telugu_output.telugu_text, lang='te')
tts_te.save('speech_te.mp3')
print('\n🔊 Telugu Speech (తెలుగు):')
display(Audio('speech_te.mp3'))
```
